### Exercice 11

```python
# Créer une fonction qui renvoie une chaîne de caractères
# à partir d'une liste qui contient des caractères
l = ['coucou', 'je', 'suis', 'une','tortue']

# output => 'coucou je suis une tortue'

# Solution 1
def renvoyer_chaine(l):
    return " ".join(l)

x = renvoyer_chaine(l)
print(x)

# Solution 2
def renvoyer_chaine(l):
    txt = ""
    for elem in l:
        txt += elem + " "
    return txt

print(renvoyer_chaine(l))
```

### Exercice 12

```python
# Usine de t-shirts qui produit 3 tailles 'S', 'M', 'L'
# Créer une fonction qui permet à partir d'une liste des t-shirts produits
# e.g. ['S', 'S', 'M', 'S', 'L', 'S', 'M']
# Renvoyer pour chaque taille de t-shirt, le nombre de t-shirts produits
# {'S': 2000, 'M': 5000, 'L': 2000}

# Option 1
def count_production(l):
    sizes = {'S': 0, 'M': 0, 'L': 0}
    for size in sizes:
        sizes[size] = l.count(size)
    return sizes

l = ['S', 'S', 'M', 'S', 'L', 'S', 'M']
print(count_production(l))

# Option 2
def count_production(l):
    """
    Dans cette approche on sépare la liste
    des tailles de l'objet à retourner
    """
    sizes = ['S', 'M', 'L']  
    production = {}
    for size in sizes:
        production[size] = l.count(size)
    return production

l = ['S', 'S', 'M', 'S', 'L', 'S', 'M']
print(count_production(l))

# Option 3
def count_production(l):
    """
    Idem sauf qu'on utilise la fonction set pour obtenir
    l'ensemble des tailles possibles (on obtient les 'valeurs uniques')
    """
    sizes = set(l)
    production = {}
    for size in sizes:
        production[size] = l.count(size)
    return production

l = ['S', 'S', 'M', 'S', 'L', 'S', 'M']
print(count_production(l))


# Option 4 (on ne connait pas count)
def count_production(l):
    sizes = {'S': 0, 'M': 0, 'L': 0}
    for size in sizes:
        repeats = 0
        for val in l:
            if val == size:
                repeats += 1
        sizes[size] = repeats
    return sizes

l = ['S', 'S', 'M', 'S', 'L', 'S', 'M']
print(count_production(l))
```
